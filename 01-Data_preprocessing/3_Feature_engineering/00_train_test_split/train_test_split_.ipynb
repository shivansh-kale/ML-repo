# sklearn.model_selection.train_test_split

---

## What is `train_test_split`?
`train_test_split` is a utility function from **scikit-learn** used to split data into
training and testing subsets in a **random and reproducible** way.

It helps evaluate how well a model generalizes to unseen data.

---

## Import Statement

### Best Practices

- * Always split before preprocessing

- * Use stratify for classification

- * Fix random_state during experiments

- * Never touch test data during tuning

### Common Mistakes
- ❌ Scaling before split
- ❌ Feature selection before split
- ❌ Using test data for hyperparameter tuning
- ❌ Forgetting stratification in imbalanced datasets


---



### Relation to Validation & Cross-Validation
|Technique|Role |
|-----------|-----------|
|train_test_split|Creates test set|
|Validation split|Manual split from training data|
|Cross-validation|Replaces validation set, not test set|

### Final Thoughts 

- train_test_split is the first safety step in machine learning
 
- Correct splitting prevents data leakage

- Honest evaluation starts with a clean split

📌 A model trained on clean splits is a model you can trust.
