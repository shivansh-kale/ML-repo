### 7. View vs copy

```python
a = np.arange(10)
b = a[2:5]   # view
b[:] = 99
print(a)    # original changes

#force copy
c = a[2:5].copy()
c[:] = 55
print(a)    # original unchanged
```

```python
#1.Feature Normalization
X = np.random.rand(100, 3)

m = X.mean(axis=0)      #feature
print("mean",m)
sigma = X.std(axis=0)
print("std",sigma)

X_norm = (X - m)/ sigma
```

```python
#Linear regression prediction
def predict(X, w, b):
    return X @ w + b
```

```python
#Mini batch Sampling
def get_minibatch(X, y, batch_size=32):
    idx = np.random.choice(len(X), batch_size, replace=False)
    return X[idx], y[idx]
```
