### 1.Feature space

```python
import plotly.express as px
import numpy as np
import pandas as pd

np.random.seed(0)
df = pd.DataFrame({
    "x1": np.random.randn(200),
    "x2": np.random.randn(200),
    "x3": np.random.randn(200),
    "label": np.random.choice(["A", "B"], size=200)
})

fig = px.scatter_3d(
    df, x="x1", y="x2", z="x3",
    color="label",
    title="3D Feature Space Visualization"
)
fig.show()
```

## 2.PCA projectionn

```python
from sklearn.decomposition import PCA
import plotly.express as px

X = np.random.randn(300, 5)
y = np.random.choice([0, 1], size=300)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

df_pca = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "label": y
})

fig = px.scatter(
    df_pca, x="PC1", y="PC2", color="label",
    title="PCA Projection (2D)"
)
fig.show()
```

### 3.Decision Boundry

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
import numpy as np
import matplotlib.pyplot as plt

X, y = make_classification(n_features=2, n_redundant=0, n_clusters_per_class=1)
clf = LogisticRegression().fit(X, y)

xx, yy = np.meshgrid(
    np.linspace(X[:,0].min()-1, X[:,0].max()+1, 200),
    np.linspace(X[:,1].min()-1, X[:,1].max()+1, 200)
)

Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Decision Boundary (Logistic Regression)")
plt.show()
```

## 4.Residuals vs Predictions

```python
y_true = np.random.randn(200)
y_pred = y_true + np.random.normal(0, 0.5, size=200)

residuals = y_true - y_pred

plt.scatter(y_pred, residuals)
plt.axhline(0)
plt.xlabel("Predicted")
plt.ylabel("Residuals")
plt.title("Residuals vs Predictions")
plt.show()
```

### 5.Correlation Network Heatmap

```python
import seaborn as sns

df_corr = pd.DataFrame(np.random.randn(300, 8), columns=[f"F{i}" for i in range(8)])
corr = df_corr.corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Feature Correlation Matrix")
plt.show()
```

### 6.Interactive Parallel Coordinates

```python
import plotly.express as px

df = px.data.iris()

fig = px.parallel_coordinates(
    df,
    color="species_id",
    title="Parallel Coordinates Plot (Multi-Feature View)"
)
fig.show()
```

### 7.Animated Training Loss

```python
import numpy as np
import matplotlib.pyplot as plt

epochs = np.arange(1, 51)
train_loss = np.exp(-epochs/10) + np.random.rand(50)*0.05
val_loss = train_loss + 0.1

plt.plot(epochs, train_loss, label="Train Loss")
plt.plot(epochs, val_loss, label="Val Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.show()
```
