
### 1. Core Objects
Series — 1D labeled array

DataFrame — 2D table (rows + columns)

### 2. Loading Data

`pd.read_csv()` — load CSV

`pd.read_excel()`  — load Excel

`pd.read_json()`  — load JSON

### 3. Quick Inspection

`df.head()` — first rows

`df.tail()` — last rows

`df.shape` — (rows, cols)

`df.info()` — dtypes + nulls

`df.describe()` — stats summary

### 4. Column & Row Selection

`df["col"]` — single column

`df[["c1","c2"]]` — multiple columns

`df.iloc[i, j]` — integer position

`df.loc[row, col]` — label based

### 5. Filtering Rows

`df[df["col"] > value]`

Multiple conditions:  `df[(df["A"] > 5) & (df["B"] < 10)]`

### 6. Missing Values (VERY COMMON)

`df.isnull().sum()` — count nulls

`df.dropna()` — drop rows with null

`df.fillna(value)` — fill nulls

Used for:
    -Data cleaning

    -Preprocessing

### 7. Type Conversion

`df["col"].astype(int)`

`pd.to_datetime()`

### 8. Sorting

`df.sort_values("col")`

`df.sort_values("col", ascending=False)`

### 9. GroupBy (CORE FOR ANALYSIS)

`df.groupby("col").mean()`

`df.groupby("col")["target"].agg(["mean","count"])`

Used for:

-Aggregations

-Feature insights

### 10. Apply & Map

`df["col"].apply(func)`

`df["col"].map(dict)`

Used for:

-Feature engineering

-Transformations

### 11. Renaming & Dropping

`df.rename(columns={"old":"new"})`

`df.drop(columns=["col"])`

`df.drop(index=[0,1])`

### 12. Handling Duplicates

`df.duplicated()`

`df.drop_duplicates()`

### 13. Merging & Joining (REAL-WORLD)

`pd.merge(df1, df2, on="id")`

`df.join(df2)`

Used for:

-Combining datasets

### 14. Creating New Features

`df["new_col"] = df["A"] / df["B"]`

Core for ML feature engineering.

### 15. Categorical Data

`pd.get_dummies(df["cat"])`

astype("category")

Used for:

-Encoding features

### 16. Index Handling

`df.set_index("col")`

`df.reset_index()`

### 17. Exporting Data

`df.to_csv("file.csv", index=False)`

`df.to_excel("file.xlsx")`

### 18. Performance Tips

Avoid row-wise loops

Prefer vectorized ops

Use categorical dtype for large categorie
